```python
# đếm số lần xuất hiện của ký tự
s = "Hom nay la thu hai"
d_cnt = {}
for ci in list(s):
  if d_cnt.get(ci) is None:
    d_cnt[ci] = 1
  else:
    d_cnt[ci] = d_cnt[ci]+ 1

print(d_cnt)
```
